```python
profession_toreplace=["'Civil Engineer'", 'Architect', "'UX/UI Designer'",
       "'Digital Marketer'", "'Content Writer'",
       "'Educational Consultant'", 'Teacher', 'Manager', 'Chef', 'Doctor',
       'Lawyer', 'Entrepreneur', 'Pharmacist']

df['Profession'].replace(profession_toreplace,'working_professional',inplace=True)
```

```python
#External Factors
yes_counts_gender = df[df['Depression'] == 1]['Gender'].value_counts()

yes_counts_Academic_Pressure = df[df['Depression'] == 1]['Academic Pressure'].value_counts()

yes_counts_Academic_profession = df[df['Depression'] == 1]['Profession'].value_counts()

yes_counts_Academic_degree = df[df['Depression'] == 1]['Degree'].value_counts()

yes_counts_suicidal_thoughts= df[df['Depression']==1]['suicidal_thoughts'].value_counts()

yes_counts_Study_satisfaction= df[df['Depression']==1]['Study Satisfaction'].value_counts()

#Health Factors
yes_counts_Dietary_Habits= df[df['Depression']==1]['Dietary Habits'].value_counts()

yes_counts_sleep_duration= df[df['Depression']==1]['Sleep Duration'].value_counts()

yes_counts_Mental_Illness= df[df['Depression']==1]['Family History of Mental Illness'].value_counts()
```

```python
# External Factors
fig = plt.figure(figsize=(30, 15)) 
fig.suptitle('Depression based on External Factors',fontsize=24) 
ax1 = fig.add_subplot(2, 3, 1)  
ax2 = fig.add_subplot(2, 3, 2)  
ax3 = fig.add_subplot(2, 3, 3)  
ax4 = fig.add_subplot(2, 3, 4)
ax5 = fig.add_subplot(2, 3, 5)
ax6 = fig.add_subplot(2, 3, 6)
ax1.bar(x=yes_counts_gender.index, height=yes_counts_gender.values)
ax1.set_xlabel('Gender')
ax1.set_ylabel('Count')
ax1.set_title("Depression for each Gender")
ax2.bar(x=yes_counts_Academic_Pressure.index, height=yes_counts_Academic_Pressure.values)
ax2.set_xlabel('Academic pressure')
ax2.set_ylabel('Count')
ax2.set_title("Depression based on academic pressure")
ax3.bar(x=yes_counts_Academic_profession.index, height=yes_counts_Academic_profession.values)
ax3.set_xlabel('Profession')
ax3.set_ylabel('Count')
ax3.set_title("Depression based on Profession")
ax4.bar(x=yes_counts_Academic_degree.index, height=yes_counts_Academic_degree.values)
ax4.set_xlabel('Degree')
ax4.set_ylabel('Count')
ax4.set_title("Depression based on Degrees")
ax4.set_xticklabels(ax4.get_xticklabels(), rotation=90)
ax5.bar(x=yes_counts_Study_satisfaction.index, height=yes_counts_Study_satisfaction.values)
ax5.set_xlabel('Study Satisfaction')
ax5.set_ylabel('Count')
ax5.set_title("Depression based on Study Satisfaction")
ax6.bar(x=yes_counts_suicidal_thoughts.index, height=yes_counts_suicidal_thoughts.values)
ax6.set_xlabel('Suicidal Thoughts')
ax6.set_ylabel('Count')
ax6.set_title("Depression based on Suicidal Thoughts")
plt.tight_layout()
plt.show()

#Health Factors
fig = plt.figure(figsize=(10, 4))  
fig.suptitle('Depression based on Health Factors')
ax1 = fig.add_subplot(1, 3, 1)  
ax2 = fig.add_subplot(1, 3, 2)  
ax3 = fig.add_subplot(1, 3, 3)  
ax1.bar(x=yes_counts_Dietary_Habits.index, height=yes_counts_Dietary_Habits.values)
ax1.set_xlabel('Dietary Habits')
ax1.set_ylabel('Count')
ax1.set_title("Depression based on Diet")
ax1.set_xticklabels(ax1.get_xticklabels(),rotation=90)
ax2.bar(x=yes_counts_sleep_duration.index, height=yes_counts_sleep_duration.values)
ax2.set_xlabel('Sleep Duration')
ax2.set_ylabel('Count')
ax2.set_title("Depression based on Sleep Duration")
ax2.set_xticklabels(ax2.get_xticklabels(),rotation=90)
ax3.bar(x=yes_counts_Mental_Illness.index, height=yes_counts_Mental_Illness.values)
ax3.set_xlabel('Family History of Mental Illness')
ax3.set_ylabel('Count')
ax3.set_title("Depression based on Family History")
plt.tight_layout()
plt.show()
```

#### Analysis based on above graphs

- Males tend to go more in depression.<br>
- It seems that students who don't take pressure of studies tend to go less in depression.<br>
- If we see the profession graph, then we can see that the students tend to go in depression a lot compared to the working professionals. So for further analysis, we will focus on students and their depression. So, to make the model more accurate and reduce the computational load, we will drop the columns which are focused on profession.<br>
- After further analysis, we can also see the amount of depression is lot in Class 12 compared to other degrees or field of education. <br>
- Surprisingly, the diet also have adverse effect on depression. <br>
- And it is quite evident that if a student is having a suicidal thought, probability is he/she is in depression.<br>
- Lesser the sleep duration, higher are the chances of depression.<br>
- Family History doesn't play an important role in student going in depression.<br>
- More the Study satisfaction, lesser are the chances of going in depression. For value 0, it seems that student doesn't care about the study.<br><br>

Based on above findings and analysis, we will alter our dataset to remove the values that doesn't contribute.

```python
# we will see if the dataset is imbalanced.
counts=df['Depression'].value_counts()
ratio=np.round(counts[1]/counts[0],2)
print(ratio)
```

- The dataset is slightly imbalanced toward people who are depressed.

- But 1.41 is not extremely imbalanced — models can still handle it decently without special treatment (like SMOTE, class weighting, etc.).

- If the ratio were something like 5, 10, 20, then you would definitely need to balance the classes before training a machine learning model.
